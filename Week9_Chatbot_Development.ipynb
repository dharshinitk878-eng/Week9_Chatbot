
# Week 9 - Building an Interactive AI Chatbot Using Ollama

## 1. Project Introduction

This project is about creating an interactive AI chatbot using Python and Ollama.

The chatbot answers questions related to Artificial Intelligence, Python, Computer Science and Technology.

It can remember previous messages during a conversation and save chat history in a CSV file.

The main aim of this project is to understand how AI chatbots work.



## 2. Libraries Used

The following libraries are used in this project:

1. **Ollama:** Connects Python with the local AI model.
2. **CSV:** Saves chatbot conversations in a CSV file.
3. **Datetime:** Records the date and time of each conversation.

These libraries help the chatbot communicate with the AI model and store conversation details.



## 3. Ollama Model Setup

Ollama is a tool used to run AI models on a computer.

In this project, we use the Gemma 3 1B model.

The model is downloaded using the following command:

ollama pull gemma3:1b

Ollama allows the Python program to communicate with the AI model and generate answers to user questions.



## 4. Chatbot Implementation

The chatbot is developed using Python and Ollama.

The user can enter questions through the terminal.

The Ollama model processes the questions and generates answers.

The chatbot also maintains conversation history, so it can remember previous messages.

The conversations are saved in a CSV file with date and time.


```python

import ollama
import csv
from datetime import datetime

messages = [
    {
        "role": "system",
        "content": "You are an educational chatbot. Explain AI, Python and Computer Science in simple words."
    }
]

print("Welcome to Educational AI Chatbot!")

while True:
    question = input("You: ")

    if question.lower() == "exit":
        print("Chatbot: Goodbye!")
        break

    messages.append({
        "role": "user",
        "content": question
    })

    response = ollama.chat(
        model="gemma3:1b",
        messages=messages
    )

    answer = response["message"]["content"]

    messages.append({
        "role": "assistant",
        "content": answer
    })

    print("Chatbot:", answer)

    with open("dataset/chat_history.csv", "a", newline="", encoding="utf-8") as file:
        writer = csv.writer(file)
        writer.writerow([datetime.now(), question, answer])
```
